# Data Cleaning

Data cleaning is the process of identifying and handling issues in the dataset such as:

- Missing values
- Duplicate records
- Incorrect data types
- Unnecessary columns
- Inconsistent data

A clean dataset improves the quality of analysis and helps build better machine learning models.

```python
# to check missing values

df.isnull().sum()
```

```python
# checking the missing values in %

missing_values = pd.DataFrame({
    "Percentage": (df.isnull().sum()/len(df))*100
})

missing_values
```

### Observation

The dataset contains missing values in:

- Age
- Cabin
- Embarked

The Cabin column has a large number of missing values, while Age and Embarked contain comparatively fewer missing values.

```python
# to check duplicate rows

df.duplicated().sum()
```

### Observation

No duplicate rows were found in the dataset.

```python
# to check unique values
df.nunique()

```

### Observation

Columns such as PassengerId and Name contain unique values for almost every passenger, whereas columns like Sex, Embarked, and Pclass contain repeated categories.

```python
df.dtypes
```

```python
# checked for unique values so that to convert suitable into category

df["Survived"] = df["Survived"].astype("category")
df["Pclass"] = df["Pclass"].astype("category")
df["Sex"] = df["Sex"].astype("category")
df["Embarked"] = df["Embarked"].astype("category")
```

### Observation

Categorical columns were converted to the `category` data type. This reduces memory usage and makes the nature of the variables clearer for analysis.

```python
df.isnull().sum()
```

```python
sns.histplot(x="Age",data=df,kde=True)
```

```python
df["Age"].describe()
```

```python
df["Age"].fillna(df["Age"].median(), inplace=True)

# fiiling with median because the graph is skewed
```

```python
df["Embarked"].fillna(df["Embarked"].mode()[0], inplace=True)

#filling with mode because the its categorial
```

```python
#dropping cabin beacuse most of the value is missing and we cant predict the cabin no

df.drop("Cabin", axis=1, inplace=True)
```

### Observation

- Missing values in Age were filled using the median.
- Missing values in Embarked were filled using the mode.
- Cabin was removed due to a high percentage of missing values.

```python
df.shape
```

```python
df.info(memory_usage="deep")
```

```python
# saving the clean dataset

df.to_csv("Titanic_Cleaned.csv", index=False)
```

# Univariate Analysis

Univariate analysis focuses on analyzing one variable at a time. It helps understand the distribution, spread, frequency, and characteristics of individual features in the dataset.

The analysis is divided into:

- Numerical Features
- Categorical Features

## Numerical Features

```python
df["Age"].describe()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["Age"],
    bins=30,
    edgecolor="black"
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

### Observation

- Most passengers were between 20 and 40 years old.
- The distribution is slightly right-skewed.
- Very young and elderly passengers are fewer in number.

```python
#KDE plot

plt.figure(figsize=(8,5))

sns.kdeplot(
    df["Age"],
    fill=True
)

plt.title("Age Density Distribution")
plt.xlabel("Age")

plt.show()
```

### Observation

The KDE plot provides a smooth representation of the age distribution and confirms that most passengers were young adults.

```python
#To identify outliers

plt.figure(figsize=(8,2))

sns.boxplot(
    x=df["Age"]
)

plt.title("Age Box Plot")

plt.show()
```

### Observation

- A few elderly passengers appear as outliers.
- Most passengers fall within a moderate age range.
- The box plot helps identify the spread and potential outliers.

# Bivariate Analysis

Bivariate analysis examines the relationship between two variables. It helps identify trends, correlations, and patterns that cannot be observed by analyzing variables individually.

The analysis is divided into:

- Numerical vs Numerical
- Categorical vs Categorical
- Categorical vs Numerical

## Numerical vs Numerical

```python

df[["Age", "Fare"]].corr()
```

```python
plt.figure(figsize=(5,4))

sns.heatmap(
    df[["Age","Fare"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation between Age and Fare")

plt.show()
```

### Observation

The correlation coefficient between Age and Fare is weak, indicating that passenger age has little influence on the ticket fare.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x="Age",
    y="Fare",
    data=df
)

plt.title("Age vs Fare")

plt.show()
```

### Observation

The scatter plot does not show a strong linear relationship between Age and Fare. Most passengers paid lower fares regardless of their age.

```python
plt.figure(figsize=(8,5))

sns.regplot(
    x="Age",
    y="Fare",
    data=df,
    scatter_kws={"alpha":0.5},
    line_kws={"color":"red"}
)

plt.title("Regression Line: Age vs Fare")

plt.show()
```

### Observation

The regression line is nearly flat, confirming that there is only a weak relationship between Age and Fare.

### Summary

- Weak correlation between Age and Fare.
- No strong linear relationship.
- Age alone does not explain ticket price.

# Multivariate Analysis

Multivariate analysis examines the relationship between three or more variables simultaneously. It helps uncover deeper patterns, interactions, and trends that cannot be observed using univariate or bivariate analysis.

In this phase, we analyze multiple variables together using:

- Scatter Plot with Hue
- Box Plot with Hue
- Bar Plot with Hue
- Pair Plot
- Correlation Heatmap

## Age vs Fare vs Survived

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Age",
    y="Fare",
    hue="Survived",
    data=df
)

plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

### Observation

- Survivors are more frequently observed among passengers who paid higher fares.
- There is no strong relationship between Age and Fare.
- Fare appears to have a greater influence on survival than Age.

## Passenger Class vs Fare vs Survival

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    x="Pclass",
    y="Fare",
    hue="Survived",
    data=df
)

plt.title("Fare Distribution by Passenger Class and Survival")

plt.show()
```

### Observation

- First-class passengers generally paid higher fares.
- Survivors are more common among passengers paying higher fares, especially in first class.
- Third-class passengers mostly paid lower fares and experienced lower survival rates.

## Passenger Class vs Fare vs Gender

```python
plt.figure(figsize=(8,6))

sns.barplot(
    x="Pclass",
    y="Fare",
    hue="Sex",
    data=df
)

plt.title("Average Fare by Passenger Class and Gender")

plt.show()
```

### Observation

Within each passenger class, the average fare paid by male and female passengers is broadly similar. However, average fares differ considerably across passenger classes.

## Pair Plot

```python
sns.pairplot(
    df[
        [
            "Age",
            "Fare",
            "SibSp",
            "Parch",
            "Survived"
        ]
    ],
    hue="Survived"
)

plt.show()
```

### Observation

## Correlation Heatmap

# Feature Engineering

Feature engineering is the process of creating new features from existing data to improve the predictive power of machine learning models.

Well-designed features often improve model performance more than simply choosing a different algorithm.

In this phase, new features are created based on insights obtained during exploratory data analysis.

```python
# Create FamilySize feature

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
```

```python
df[["SibSp", "Parch", "FamilySize"]].head()
```

### Observation

The FamilySize feature represents the total number of family members travelling together, including the passenger.

This feature may provide more meaningful information than using SibSp and Parch independently.

```python
# Create IsAlone feature

df["IsAlone"] = (df["FamilySize"] == 1).astype(int)
```

```python
df[["FamilySize", "IsAlone"]].head()
```

### Observation

The IsAlone feature identifies whether a passenger travelled alone.

Passengers travelling alone may have experienced different survival outcomes compared to those travelling with family.

```python
df.head()
```

# Phase 6 Summary

The following features were created:

- **FamilySize** – Total family members travelling together.
- **IsAlone** – Indicates whether the passenger travelled alone.


These engineered features provide additional information that may improve machine learning model performance.
